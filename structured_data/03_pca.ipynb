```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

matrix = pd.read_pickle("cache/matrix.pkl")

X_train = matrix[matrix.date_block_num < 33].drop(['item_cnt_month'], axis=1)
X_valid = matrix[matrix.date_block_num == 33].drop(['item_cnt_month'], axis=1)
X_test = matrix[matrix.date_block_num == 34].drop(['item_cnt_month'], axis=1)

drop_cols = ['item_cnt_month_lag_1', 'item_cnt_month_lag_2', 
            'item_cnt_month_lag_3', 'item_cnt_month_lag_6',
            'item_cnt_month_lag_12']

X_train = X_train.drop(columns=drop_cols)
X_valid = X_valid.drop(columns=drop_cols)
X_test = X_test.drop(columns=drop_cols) 


scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_valid_scaled = scaler.transform(X_valid)
X_test_scaled = scaler.transform(X_test)

pca = PCA(n_components=0.95)
X_train_pca = pca.fit_transform(X_train_scaled)
X_valid_pca = pca.transform(X_valid_scaled)
X_test_pca = pca.transform(X_test_scaled)

print("元の次元数:", X_train.shape[1])
print("PCA後の次元数:", X_train_pca.shape[1])

print("各主成分の寄与率:", pca.explained_variance_ratio_)
print("累積寄与率:", np.cumsum(pca.explained_variance_ratio_))
print("選ばれた主成分の数:", pca.n_components_)
```
