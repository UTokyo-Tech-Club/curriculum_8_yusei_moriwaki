```python
import nltk
import gensim
from nltk.corpus import brown

nltk.download('brown')

brown.sents()

model = gensim.models.Word2Vec(brown.sents(), vector_size=100, seed=0, sg=1)

model.wv["mouth"]

print(model.wv.most_similar('mouth', topn=10))

print(model.wv.most_similar('computer', topn=10))
```
